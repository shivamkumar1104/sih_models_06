# 🚢 Ship & Vessel Classification & Dimensional Engine
### Smart India Hackathon (SIH 2024)
This notebook analyzes physical bulk carrier vessel specifications (`dataset_used_cleaned_ships.csv`) to classify vessels into Handysize, Supramax, Panamax, and Capesize.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('dataset_used_cleaned_ships.csv')
print("Ship Specs Dataset Shape:", df.shape)
df.head(10)
```

### 1. Summary Statistics by Vessel Class


```python
summary = df.groupby('vessel_class').agg({
    'deadweight': ['count', 'mean', 'min', 'max'],
    'length_m': ['mean', 'min', 'max'],
    'width_m': ['mean', 'min', 'max']
}).round(1)
print(summary)
```

### 2. Vessel Class Distribution & Dimensional Scatter Plot


```python
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
df['vessel_class'].value_counts().plot(kind='bar', color=['#06b6d4', '#8b5cf6', '#10b981', '#f59e0b'])
plt.title('Vessel Class Distribution')
plt.xlabel('Vessel Class')
plt.ylabel('Count')

plt.subplot(1, 2, 2)
sns.scatterplot(data=df, x='deadweight', y='length_m', hue='vessel_class', palette='viridis')
plt.title('DWT vs Length Overall (LOA)')
plt.tight_layout()
plt.show()
```
