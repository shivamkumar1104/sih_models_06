# ⚓ East Coast India Port Optimization & Turnaround Engine
### Smart India Hackathon (SIH 2024)
This notebook models port berth constraints (draft, LOA, beam, handling rates) for Visakhapatnam, Paradip, Gangavaram, Dhamra, Gopalpur, and Haldia.


```python
import pandas as pd
import matplotlib.pyplot as plt

port_df = pd.read_csv('port_constraints.csv')
print("East Coast Port Constraints Table:")
print(port_df.to_string(index=False))
```

### 1. Port Draft & Handling Rate Comparison


```python
plt.figure(figsize=(14, 5))
plt.subplot(1, 2, 1)
plt.bar(port_df['port_name'], port_df['max_draft_m'], color='#06b6d4')
plt.title('Max Permissible Draft (Meters)')
plt.ylabel('Draft (m)')
plt.xticks(rotation=30)

plt.subplot(1, 2, 2)
plt.bar(port_df['port_name'], port_df['handling_rate_tpd'], color='#10b981')
plt.title('Daily Cargo Handling Capacity (TPD)')
plt.ylabel('MT / Day')
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```
